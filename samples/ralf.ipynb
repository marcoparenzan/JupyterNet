# Ralf kernel sample

A `ralf` cell's text is sent as a prompt to RalfAI's own in-process agent, which gets tools (`Notebook_ListCells`/`GetCell`/`RunCell`/`SetCellCode`) to inspect, run and edit this very notebook's *other* cells. Needs an existing RalfAI configuration (`RALFAI_CONFIG_PATH` or `appsettings.json` next to `JupyterNet.Host` — see USAGE.md) — without one, the cell fails with a clear error instead of crashing the host.
