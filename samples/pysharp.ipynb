# PySharp kernel sample

A Python interpreter written from scratch in C# (`PySharpLib.PyEngine`). Cross-cell state is carried forward by copying every global out of the finished module after each cell. `pandas` (`PySharp.Pandas`, a native .NET pandas, no CPython) is imported like any other module, and a `DataFrame` auto-displays as an HTML table via its `_repr_html_`, exactly like real Jupyter.

```jupyternet
numbers = list(range(1, 11))
squares = [n * n for n in numbers]
print(f"squares = {squares}")
sum(squares)
```

```jupyternet
# `squares` is still here — carried forward after each cell.
[n for n in squares if n % 2 == 0]
```
