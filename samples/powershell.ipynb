# PowerShell kernel sample

`Microsoft.PowerShell.SDK` — a real embedded PowerShell 7+ `Runspace` per session; a fresh pipeline per cell, all against that same `Runspace`, so `$x` set in one cell is still there in the next.

```jupyternet
$numbers = 1..10
$squares = $numbers | ForEach-Object { $_ * $_ }
Write-Host "squares = $($squares -join ', ')"
($squares | Measure-Object -Sum).Sum
```

```jupyternet
# `$squares` is still here — same Runspace across cells.
$squares | Where-Object { $_ % 2 -eq 0 } | Measure-Object -Sum | Select-Object -ExpandProperty Sum
```
