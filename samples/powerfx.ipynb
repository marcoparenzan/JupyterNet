# Power Fx kernel sample

`Microsoft.PowerFx.Interpreter`'s `RecalcEngine` — a cell is **one formula**, the same grain a Power Apps formula bar works at, not a script of statements. `Set(name, expr)` is how a cell assigns a session variable; the variable is then visible, by name, to every formula cell after it.

```jupyternet
Set(price, 19.99)
```

```jupyternet
Set(quantity, 3)
```

```jupyternet
// `price` and `quantity` are still here — one RecalcEngine per session.
Round(price * quantity, 2)
```

```jupyternet
If(price * quantity > 50, "free shipping", "standard shipping")
```

```jupyternet
Concatenate("Total: $", Text(price * quantity, "0.00"))
```
