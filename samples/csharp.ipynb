# C# kernel sample

Roslyn scripting (`Microsoft.CodeAnalysis.CSharp.Scripting`) — a REPL-style C#. Top-level variables declared in one cell stay visible in the next; `Display.Html` pushes rich output explicitly.

```jupyternet
var numbers = Enumerable.Range(1, 10).ToArray();
var squares = numbers.Select(n => n * n).ToArray();
Display.Html($"<p>squares = [{string.Join(", ", squares)}]</p>");
squares.Sum()
```

```jupyternet
// `squares` is still here — session state persists across cells.
squares.Where(s => s % 2 == 0).Sum()
```
