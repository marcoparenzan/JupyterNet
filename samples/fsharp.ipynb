# F# kernel sample

`FSharp.Compiler.Service`'s `FsiEvaluationSession` — a real, long-lived FSI session, so cross-cell state persists on its own, no workaround needed.

```jupyternet
let numbers = [1..10]
let squares = numbers |> List.map (fun n -> n * n)
printfn "squares = %A" squares
List.sum squares
```

```jupyternet
// `squares` is still here — a real FSI session.
squares |> List.filter (fun n -> n % 2 = 0) |> List.sum
```

```jupyternet
type Shape =
    | Circle of radius: float
    | Rectangle of width: float * height: float

let area shape =
    match shape with
    | Circle r -> System.Math.PI * r * r
    | Rectangle (w, h) -> w * h

area (Circle 2.0)
```
