# Ontly kernel sample

A cell is one full, standalone YAML domain contract — compiled with the real `Ontly.Domain.ContractCompiler` against both the C# and Python generators. The output is a self-contained HTML report: a summary table plus the generated code side by side.

```jupyternet
context: catalog
namespace: JupyterNet.Catalog
scalars:
  Sku:
    base: string
    maxLength: 40
  Money:
    base: decimal
types:
  Product:
    kind: dto
    attributes:
      sku: Sku
      price: Money
```
